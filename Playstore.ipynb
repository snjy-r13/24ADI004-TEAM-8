```python
df["Size"] = df["Size"].replace("Varies with device", np.nan)
```

```python
def convert_size(size):
    if pd.isnull(size):
        return np.nan
    
    size = str(size).replace(",", "").strip()  
    
    if size.endswith("M"):
        return float(size.replace("M", ""))
    
    elif size.endswith("k"):
        return float(size.replace("k", "")) / 1000
    
    else:
        return np.nan

df["Size"] = df["Size"].apply(convert_size)
```

```python
num_cols = ['Rating', 'Rating Count', 'Installs',
            'Minimum Installs', 'Maximum Installs',
            'Price', 'Minimum Android']

for col in num_cols:
    plt.figure(figsize=(4,2))
    
    # Apply log for highly skewed columns
    if col in ['Rating Count', 'Installs', 'Minimum Installs', 'Maximum Installs']:
        sns.boxplot(x=np.log1p(df[col]))
        plt.title(f"Boxplot of Log({col})")
    else:
        sns.boxplot(x=df[col])
        plt.title(f"Boxplot of {col}")
        
    plt.show()
```

```python
from scipy import stats
for col in num_cols:
    
    # Use log transform for skewed columns
    if col in ['Rating Count', 'Installs', 'Minimum Installs', 'Maximum Installs']:
        data = np.log1p(df[col])
    else:
        data = df[col]
    
    z_scores = np.abs(stats.zscore(data))
    outliers = np.sum(z_scores > 3)
    
    print(f"{col} - Number of Outliers (Z>3): {outliers}")
```

UNIVARIATE ANALYSIS

**EDA**

Rating Distribution (Histogram + KDE)

```python
plt.figure()
sns.histplot(df['Rating'], bins=30, kde=True)
plt.title("Distribution of App Ratings")
plt.xlabel("Rating")
plt.ylabel("Frequency")
plt.show()
```

```python
df_valid = df[df['Rating'] > 0]

sns.histplot(df_valid['Rating'], bins=30, kde=True)
plt.title("Distribution of App Ratings (Excluding 0 Ratings)")
plt.show()
```

```python
# Remove invalid ratings if not already done
df_valid = df[df['Rating'] > 0].copy()

# Create log transformed installs
df_valid['Log_Installs'] = np.log1p(df_valid['Installs'])
```

Price Distribution (Paid Apps Only)


```python
# Create paid apps dataframe
paid_apps = df[df['Free'] == False].copy()
plt.figure()
sns.histplot(np.log1p(paid_apps['Price']), bins=30, kde=True)
plt.title("Log Transformed Price Distribution")
plt.xlabel("Log(Price)")
plt.show()
```

App Category Count (Top 15)

```python
top_categories = df['Category'].value_counts().head(15)

plt.figure()
sns.barplot(x=top_categories.values, y=top_categories.index)
plt.title("Top 15 App Categories")
plt.xlabel("Number of Apps")
plt.show()
```

BIVARIATE ANALYSIS

Rating vs Installs (Log Scale Scatter)

```python
sns.regplot(
    x=np.log1p(df_valid['Installs']),
    y=df_valid['Rating'],
    scatter_kws={'alpha':0.2}
)
plt.title("Rating vs Installs with Trend Line")
plt.show()
```

Free vs Paid Rating Comparison (Boxplot)

```python
df_valid = df[df['Rating'] > 0]

plt.figure()
sns.boxplot(x='Free', y='Rating', data=df_valid)
plt.title("Free vs Paid Apps Rating Comparison (Valid Ratings Only)")
plt.show()
```

```python
df_valid.groupby('Free')['Rating'].agg(['mean','median','std'])
```

```python
numeric_df = df.select_dtypes(include=np.number)

plt.figure(figsize=(12,8))
sns.heatmap(numeric_df.corr(), annot=True, cmap='coolwarm', fmt=".2f")
plt.title("Correlation Matrix")
plt.show()
```

```python
top5 = df['Category'].value_counts().head(5).index
df_top = df[df['Category'].isin(top5)]

plt.figure(figsize=(12,6))
sns.boxplot(x='Category', y=np.log1p(df_top['Installs']), data=df_top)
plt.xticks(rotation=45)
plt.title("Install Distribution by Top Categories (Log Scale)")
plt.show()
```

```python
category_rating = df_valid.groupby('Category')['Rating'].mean().sort_values(ascending=False).head(10)

plt.figure(figsize=(10,6))
sns.barplot(x=category_rating.values, y=category_rating.index)
plt.title("Top 10 Categories by Average Rating")
plt.xlabel("Average Rating")
plt.show()
```
