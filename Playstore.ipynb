```python
def convert_size(size):
    if pd.isnull(size):
        return np.nan
    
    size = str(size).replace(",", "").strip()  
    
    if size.endswith("M"):
        return float(size.replace("M", ""))
    
    elif size.endswith("k"):
        return float(size.replace("k", "")) / 1000
    
    else:
        return np.nan

df["Size"] = df["Size"].apply(convert_size)
```

Drop Unnecessary Columns

```python
df = df.drop(columns=[
    'App Name', 'App Id',
    'Developer Id', 'Developer Email'
], errors='ignore')
```

Handle Missing Values

```python
# Remove rows without ratings
df = df.dropna(subset=['Rating'])

# Fill numeric missing values
df['Size'].fillna(df['Size'].median(), inplace=True)
df['Price'].fillna(0, inplace=True)
df['Rating Count'].fillna(0, inplace=True)
```

Fix Skewness

```python

skew_cols = ['Installs', 'Minimum Installs', 'Maximum Installs', 'Rating Count', 'Price']

for col in skew_cols:
    df[col] = np.log1p(df[col])
```

Feature Engineering

```python
# App age
df['App Age'] = df['Update Year'] - df['Release Year']

# Recently updated flag
df['Is Recently Updated'] = (df['Update Year'] >= 2020).astype(int)
```

```python
from sklearn.preprocessing import LabelEncoder

le = LabelEncoder()

df['Category'] = le.fit_transform(df['Category'].astype(str))
df['Content Rating'] = le.fit_transform(df['Content Rating'].astype(str))
```

Convert Boolean to Int

```python
bool_cols = ['Free', 'Ad Supported', 'In App Purchases', 'Editors Choice']

for col in bool_cols:
    df[col] = df[col].astype(int)
```

```python
# Already datetime → extract useful info
df['Release Month'] = df['Released'].dt.month
df['Update Month'] = df['Last Updated'].dt.month

# Drop original datetime columns
df = df.drop(columns=['Released', 'Last Updated'])
```

```python
df = df.drop(columns=['Minimum Installs', 'Maximum Installs'], errors='ignore')
```

**WEEK - 7**

```python
features = [
    'Category', 'Rating Count', 'Installs', 'Price', 'Size',
    'Minimum Android', 'Free', 'Content Rating',
    'Ad Supported', 'In App Purchases', 'Editors Choice',
    'App Age', 'Is Recently Updated',
    'Release Month', 'Update Month'
]

X = df[features]
```

```python
from sklearn.preprocessing import StandardScaler

scaler = StandardScaler()
X_scaled = scaler.fit_transform(X)
```

PCA

```python
from sklearn.decomposition import PCA

pca = PCA()
X_pca = pca.fit_transform(X_scaled)
```

Explained Variance

```python
plt.figure(figsize=(8,5))
plt.plot(np.cumsum(pca.explained_variance_ratio_), marker='o')
plt.xlabel("Number of Components")
plt.ylabel("Cumulative Explained Variance")
plt.title("PCA Explained Variance")
plt.grid()
plt.show()
```

Finding:
The cumulative explained variance increases rapidly and reaches around 95% with ~10–11 components, after which the curve flattens.

Insight:
Most of the important information in the dataset can be captured using fewer components, indicating high redundancy and multicollinearity among features. PCA effectively reduces dimensionality without significant information loss.

Reduce Dimensions (95%)

```python
pca = PCA(n_components=0.90)
X_pca = pca.fit_transform(X_scaled)

print("Original shape:", X_scaled.shape)
print("Reduced shape:", X_pca.shape)
```

```python
pca_2 = PCA(n_components=2)
X_pca_2 = pca_2.fit_transform(X_scaled)

plt.figure(figsize=(7,5))
plt.scatter(X_pca_2[:,0], X_pca_2[:,1], alpha=0.5)
plt.xlabel("Principal Component 1")
plt.ylabel("Principal Component 2")
plt.title("PCA Visualization")
plt.show()
```
